```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

#dados gerados

np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + 1 + 0.1 * np.random.randn(100)

#grafico de dispersao com ajuste de regressao linear

sns.regplot(x=x, y=y,)

#plot

plt.show()
```

#### boxplot

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

dias = np.random.choice(['segunda', 'Sexta', 'terça', 'Domingo'], size=100)
total_pagar = np.random.uniform(20,5,size= 100)

#grafico do tipo bloxplot

sns.boxplot(x=dias, y=total_pagar)
plt.show()
```

#### Histograma

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x = np.random.seed(42)
total_pagar = np.random.normal(30,10,size= 200)

#grafico do tipo histograma com estimativa de Kernel

sns.histplot(total_pagar, kde=True)
plt.show()
```

#### Mapa de calor

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x = np.random.seed(42)
data = np.random.rand(12,5)

#grafico do tipo heatmap

sns.heatmap(data,annot=True, cmap='YlGnBu')
plt.show()
```

#### Pair Plot

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x = np.random.seed(42)
iris = sns.load_dataset('iris')

#Mapa de calor
sns.pairplot(iris, hue='species')
plt.show()
```
